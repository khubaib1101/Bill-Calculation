```python
#  Shopping discount 8
def calculate_bill(amount,discount_type='normal'):
  org_price = amount
  print("Your original amount is: ",org_price)
  if discount_type  == 'normal':
    if amount>=10000:
      discount=amount * 0.20
    elif amount>=5000 and amount<=9999:
      discount=amount * 0.10
    elif amount>=2000 and amount<=4999:
      discount=amount * 0.05
    elif amount<2000:
      discount = 0
  print("Your discount is: ",discount)
  final_price = org_price - discount
  print("Your final price is: ",final_price)
calculate_bill(10000)


```
